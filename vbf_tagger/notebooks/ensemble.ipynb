```python
def get_mlp_predictions(model, loader):
    preds, targets = [], []
    model.eval()
    with torch.no_grad():
        for X, y in loader:
            logits = model(X)
            probs = torch.sigmoid(logits).squeeze()
            preds.append(probs.cpu().numpy())
            targets.append(y.cpu().numpy())
    return np.concatenate(preds), np.concatenate(targets)
```
